```python
data.duplicated().sum()        # to detect the all duplicate rows in the data
data.duplicated().value_counts()      # above command will only count duplicate row but this command also count non duplicate rows too

data.drop_duplicates( inplace = True )      # to delete the duplicate rows 
```

```python
data.isna().sum()                 # to count the all values in the 
# data.isna().value_counts()

data.drop( ['license', 'house_rules'] ,axis = 1 ,  inplace = True )
(data.isna().sum() / len ( data) )* 100             #  to show that how many percent data is messing at every columns 
```

```python
data.dropna(inplace = True)          # to delete all rows with any null values
```

```python
data['host_identity_verified'].str.upper()      # this command will show all the values in perticular column with 'string' data type in uppercase formate 
data['host_identity_verified'] = data['host_identity_verified'].str.upper()      # this command will replace the all values in given columns into the newly generated values from upper command 
```

```python
data['instant_bookable'] = data['instant_bookable'].apply(lambda x : 1 if x == True else 0 ) 
 # in column called 'instant_bookable' we have a vlaues as ' True and False' and this command change those values into 1 and 0 
```

```python
type(data['instant_bookable'][5])           # to check the data type of the 5th no row variable of'instant_bookable' column
data['instant_bookable'][5].dtype           # this command works similar
data['instant_bookable'].dtype          # for whole columns 
```

```python
data.reset_index(inplace = True )    # this command is for reseting the index value of the table ( because when we delete the null value rows in table through dropna() then it also delete the index value with it ) 
```

```python
data['price'] = data['price'].str.replace('$', '')      # this command will remove the '$' symbol but because of .str command the result will be in string format
                                        # .str = replace and some other commands are not works on interger so .str change it into string and after other task can be perform easily
data['price']=data['price'].str.replace('$','').astype(int64)  # this command will change all values into intger after performing the task ( but this command will show an error if non integer are presints in the columns) 
data['price'] = pd.to_numeric(data['price'].str.replace('$', ''), errors='coerce')        # this is the perfenct command to perform this task
```
